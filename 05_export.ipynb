```python
MY_ID = "arzoo"
VOL = f"/Volumes/workspace/capstone_{MY_ID}/raw"
S   = f"workspace.capstone_{MY_ID}"

dbutils.fs.mkdirs(f"{VOL}/export")

def export(tbl, name, cols=None):
    df = spark.table(f"{S}.{tbl}")
    if cols: df = df.select(*cols)
    tmp = f"{VOL}/export/_tmp_{name}"
    df.coalesce(1).write.mode("overwrite").option("header", True).csv(tmp)
    part = [f.path for f in dbutils.fs.ls(tmp) if f.name.startswith("part-") and f.name.endswith(".csv")][0]
    dbutils.fs.rm(f"{VOL}/export/{name}.csv")
    dbutils.fs.cp(part, f"{VOL}/export/{name}.csv")
    dbutils.fs.rm(tmp, True)

export("gold_grievance_weekly", "gold_grievance_weekly")
export("silver_complaint", "silver_complaint",
       ["complaint_id", "ward_id", "zone", "category", "status", "raised_ts",
        "closed_ts", "closure_known", "resolution_hours", "sla_breach"])

display(dbutils.fs.ls(f"{VOL}/export"))
```
